# Part A — Recursion

## A1. Define recursion and distinguish it from iteration. Explain the role of a base case and why every recursive call chain must eventually reach one.

**Answer:**

- **Recursion** is a programming technique in which a function calls itself to solve a smaller version of the same problem.
- **Iteration** repeats a set of statements using loops such as `for` or `while`.
- A **base case** is the stopping condition. It returns an answer without making another recursive call.
- Every recursive call chain must reach a base case; otherwise, calls continue indefinitely until Python raises a `RecursionError`.

**Difference:** Recursion uses function calls and the call stack, while iteration uses loop control. Iteration often uses less memory.

## A2. Write recursive Python functions for factorial, sum of the first n natural numbers, and computation of x raised to a non-negative integer power. State preconditions and base cases.

**Answer:**

Preconditions:
- Factorial: `n` must be a non-negative integer.
- Sum: `n` must be a non-negative integer.
- Power: exponent `n` must be a non-negative integer.

Base cases:
- `factorial(0) = 1`
- `natural_sum(0) = 0`
- `power(x, 0) = 1`

````python
def factorial(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n == 0:
        return 1
    return n * factorial(n - 1)


def natural_sum(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n == 0:
        return 0
    return n + natural_sum(n - 1)


def power(x, n):
    if n < 0:
        raise ValueError("Exponent must be non-negative")
    if n == 0:
        return 1
    return x * power(x, n - 1)


print(factorial(5))     # 120
print(natural_sum(5))   # 15
print(power(2, 4))      # 16
````

## A3. Design a recursive algorithm to reverse a sequence in place using start and stop indices. Trace its operation for a sequence of six elements.

**Answer:** Swap the first and last elements, then recursively process the smaller inner sequence. Stop when `start >= stop`.

For `[1, 2, 3, 4, 5, 6]`:
1. Swap indices 0 and 5 → `[6, 2, 3, 4, 5, 1]`
2. Swap indices 1 and 4 → `[6, 5, 3, 4, 2, 1]`
3. Swap indices 2 and 3 → `[6, 5, 4, 3, 2, 1]`
4. The indices meet/cross, so recursion stops.

````python
def reverse_in_place(data, start, stop):
    if start >= stop:
        return
    data[start], data[stop] = data[stop], data[start]
    reverse_in_place(data, start + 1, stop - 1)


items = [1, 2, 3, 4, 5, 6]
reverse_in_place(items, 0, len(items) - 1)
print(items)
````

## A4. Write a recursive function that checks whether a sequence is sorted in non-decreasing order. State its best- and worst-case running time.

**Answer:** A sequence is non-decreasing if every element is less than or equal to the next element. Compare adjacent elements recursively. An empty or one-element sequence is sorted.

- Best case: **O(1)** if the first pair is out of order.
- Worst case: **O(n)** if all adjacent pairs must be checked.
- Recursive call-stack space: **O(n)** in the worst case.

````python
def is_sorted(data, index=0):
    if index >= len(data) - 1:
        return True
    if data[index] > data[index + 1]:
        return False
    return is_sorted(data, index + 1)


print(is_sorted([1, 2, 2, 5]))  # True
print(is_sorted([1, 4, 3, 5]))  # False
````

## A5. Write the naive recursive Fibonacci algorithm and draw its recursion tree for n=5. Identify repeated subproblems.

**Answer:** Fibonacci is defined by `F(0)=0`, `F(1)=1`, and `F(n)=F(n-1)+F(n-2)`.

Repeated subproblems include `fib(3)` (twice) and `fib(2)` (three times). The same values are recalculated many times.

````python
def fib(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)


print(fib(5))  # 5
````

## A7. Develop a recursive algorithm for the Tower of Hanoi problem for n disks. State the three recursive steps and the base case.

**Answer:**

To move `n` disks from source to destination using an auxiliary peg:
1. Move `n-1` disks from source to auxiliary.
2. Move the largest disk from source to destination.
3. Move `n-1` disks from auxiliary to destination.

Base case: if `n == 1`, move the single disk directly. The minimum number of moves is `2^n - 1`.

````python
def tower_of_hanoi(n, source, auxiliary, destination):
    if n <= 0:
        return
    if n == 1:
        print(f"Move disk 1 from {source} to {destination}")
        return
    tower_of_hanoi(n - 1, source, destination, auxiliary)
    print(f"Move disk {n} from {source} to {destination}")
    tower_of_hanoi(n - 1, auxiliary, source, destination)


tower_of_hanoi(3, "A", "B", "C")
````

## A8. Explain tail recursion. Identify whether factorial, linear_sum, recursive binary search and recursive reversal are tail-recursive, with reasons.

**Answer:** A function is **tail-recursive** when its recursive call is the final operation; no computation remains after that call returns.

| Function | Tail-recursive? | Reason |
|---|---|---|
| Ordinary factorial `n * factorial(n-1)` | No | Multiplication remains after the call |
| Ordinary `linear_sum(n) = n + linear_sum(n-1)` | No | Addition remains after the call |
| Recursive binary search | Yes, in its usual single-branch form | It directly returns the recursive call's result |
| Recursive reversal using swaps | Yes, if the recursive call is the last operation | The swap is completed before the call |

Python does not generally optimize tail recursion, so tail-recursive Python functions still use call-stack space.

## A9. Convert a tail-recursive sequence-processing algorithm into an iterative version. Compare the use of call-stack space in both versions.

**Answer:** Example: calculate the sum of a sequence. The recursive version carries an accumulator and makes the recursive call last. The iterative version uses a loop.

- Tail-recursive version: **O(n)** call-stack space in Python.
- Iterative version: **O(1)** extra space.
- Both take **O(n)** time.

````python
def recursive_sum(data, index=0, total=0):
    if index == len(data):
        return total
    return recursive_sum(data, index + 1, total + data[index])


def iterative_sum(data):
    total = 0
    for value in data:
        total += value
    return total


numbers = [2, 4, 6, 8]
print(recursive_sum(numbers))
print(iterative_sum(numbers))
````

## C1. Define the stack ADT and specify push, pop, top, is_empty and len operations, including behaviour on an empty stack.

**Answer:** A **stack** is an abstract data type that follows LIFO (Last In, First Out).

- `push(x)`: adds `x` to the top.
- `pop()`: removes and returns the top element.
- `top()`: returns the top element without removing it.
- `is_empty()`: returns `True` if the stack has no elements.
- `len()`: returns the number of elements.

Calling `pop()` or `top()` on an empty stack should report an error, commonly `IndexError`.

## C2. Implement a Stack class using a Python list. Use a suitable exception for pop() and top() on an empty stack.

````python
class Stack:
    def __init__(self):
        self._data = []

    def push(self, value):
        self._data.append(value)

    def pop(self):
        if self.is_empty():
            raise IndexError("Pop from an empty stack")
        return self._data.pop()

    def top(self):
        if self.is_empty():
            raise IndexError("Top of an empty stack")
        return self._data[-1]

    def is_empty(self):
        return len(self._data) == 0

    def __len__(self):
        return len(self._data)


stack = Stack()
stack.push(10)
stack.push(20)
print(stack.top())
print(stack.pop())
print(len(stack))
````

## C4. Explain how a stack can reverse a sequence and trace the operations for a sample input.

**Answer:** Push every element onto the stack, then pop all elements. Since a stack is LIFO, the removal order is the reverse of the insertion order.

For input `[A, B, C]`: push A, push B, push C; then pop C, pop B, pop A. Output: `[C, B, A]`.

````python
def reverse_using_stack(sequence):
    stack = []
    for item in sequence:
        stack.append(item)

    result = []
    while stack:
        result.append(stack.pop())
    return result


print(reverse_using_stack(["A", "B", "C"]))
````

## C5. Discuss how the stack-based approach differs from counting opening and closing delimiters without preserving their order.

**Answer:** A counter can check whether the total number of opening and closing delimiters is balanced, but it cannot detect incorrect nesting. A stack stores the order of opening delimiters and checks that each closing delimiter matches the most recent unmatched opening delimiter. Thus, a stack correctly validates nested expressions such as `([{}])`.

````python
def delimiters_balanced(expression):
    pairs = {')': '(', ']': '[', '}': '{'}
    opening = set(pairs.values())
    stack = []

    for char in expression:
        if char in opening:
            stack.append(char)
        elif char in pairs:
            if not stack or stack.pop() != pairs[char]:
                return False

    return len(stack) == 0


print(delimiters_balanced("([{}])"))  # True
print(delimiters_balanced("([)]"))    # False
````

## C8. Design a circular-array queue using data, size and front variables. Derive the index formula for the next insertion and the next front position.

**Answer:** Let the array capacity be `N`, current front index be `front`, and number of elements be `size`.

- Rear insertion index: `(front + size) % N`
- Next front after dequeue: `(front + 1) % N`
- Enqueue is possible when `size < N`.
- Dequeue is possible when `size > 0`.

The modulo operation wraps indices back to zero at the end of the array.

````python
class CircularQueue:
    def __init__(self, capacity):
        if capacity <= 0:
            raise ValueError("Capacity must be positive")
        self.data = [None] * capacity
        self.capacity = capacity
        self.size = 0
        self.front = 0

    def enqueue(self, value):
        if self.size == self.capacity:
            raise IndexError("Queue is full")
        index = (self.front + self.size) % self.capacity
        self.data[index] = value
        self.size += 1

    def dequeue(self):
        if self.size == 0:
            raise IndexError("Queue is empty")
        value = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % self.capacity
        self.size -= 1
        return value

    def first(self):
        if self.size == 0:
            raise IndexError("Queue is empty")
        return self.data[self.front]

    def is_empty(self):
        return self.size == 0

    def __len__(self):
        return self.size


q = CircularQueue(3)
q.enqueue("A")
q.enqueue("B")
print(q.dequeue())
q.enqueue("C")
q.enqueue("D")
print(q.first())
````

## C10. Develop algorithms for add_first, add_last, delete_first and delete_last using circular-array indexing.

**Answer:** For capacity `N`, front index `f`, and size `s`:
- `add_first(x)`: set `f = (f - 1) % N`, store `x` at `data[f]`, increase `s`.
- `add_last(x)`: store `x` at `(f + s) % N`, increase `s`.
- `delete_first()`: read `data[f]`, set `f = (f + 1) % N`, decrease `s`.
- `delete_last()`: read and clear `data[(f + s - 1) % N]`, decrease `s`.

Each operation takes **O(1)** time when the deque has available capacity.

````python
class CircularDeque:
    def __init__(self, capacity):
        if capacity <= 0:
            raise ValueError("Capacity must be positive")
        self.data = [None] * capacity
        self.capacity = capacity
        self.front = 0
        self.size = 0

    def add_first(self, value):
        if self.size == self.capacity:
            raise IndexError("Deque is full")
        self.front = (self.front - 1) % self.capacity
        self.data[self.front] = value
        self.size += 1

    def add_last(self, value):
        if self.size == self.capacity:
            raise IndexError("Deque is full")
        index = (self.front + self.size) % self.capacity
        self.data[index] = value
        self.size += 1

    def delete_first(self):
        if self.size == 0:
            raise IndexError("Deque is empty")
        value = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % self.capacity
        self.size -= 1
        return value

    def delete_last(self):
        if self.size == 0:
            raise IndexError("Deque is empty")
        index = (self.front + self.size - 1) % self.capacity
        value = self.data[index]
        self.data[index] = None
        self.size -= 1
        return value


dq = CircularDeque(4)
dq.add_first(2)
dq.add_first(1)
dq.add_last(3)
print(dq.delete_first())
print(dq.delete_last())
````

## D3. Develop algorithms for insert_first, insert_last, delete_first and traversal. State the preconditions and pointer updates for each.

**Answer:**
- `insert_first(x)`: create a node whose `next` is the old head; update head. If the list was empty, also update tail.
- `insert_last(x)`: create a node with `next=None`; link the old tail to it, then update tail. If empty, set both head and tail.
- `delete_first()`: require a non-empty list; move head to `head.next`. If it becomes empty, set tail to `None`.
- `traversal`: start at head and follow `next` until `None`.

Insertion at either end is O(1) when both head and tail are maintained. Deleting the first node is O(1). Traversal is O(n).

````python
class SinglyLinkedList:
    class Node:
        def __init__(self, element, next_node=None):
            self.element = element
            self.next = next_node

    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def insert_first(self, value):
        self.head = self.Node(value, self.head)
        if self.size == 0:
            self.tail = self.head
        self.size += 1

    def insert_last(self, value):
        new_node = self.Node(value)
        if self.size == 0:
            self.head = new_node
        else:
            self.tail.next = new_node
        self.tail = new_node
        self.size += 1

    def delete_first(self):
        if self.size == 0:
            raise IndexError("List is empty")
        value = self.head.element
        self.head = self.head.next
        self.size -= 1
        if self.size == 0:
            self.tail = None
        return value

    def traversal(self):
        current = self.head
        result = []
        while current is not None:
            result.append(current.element)
            current = current.next
        return result


linked = SinglyLinkedList()
linked.insert_first(20)
linked.insert_first(10)
linked.insert_last(30)
print(linked.traversal())
print(linked.delete_first())
print(linked.traversal())
````

## D4. For a doubly linked list, specify the pointer updates required to insert a node between two existing nodes and to delete a node.

**Answer:** Each node has `prev` and `next` references.

To insert `x` between nodes `A` and `B`:
1. Set `x.prev = A` and `x.next = B`.
2. Set `A.next = x`.
3. Set `B.prev = x`.

To delete node `x`:
1. Set `x.prev.next = x.next`, if a previous node exists.
2. Set `x.next.prev = x.prev`, if a next node exists.
3. Update head or tail if `x` is at an end.

Boundary cases (head, tail and empty list) must be handled separately.

````python
class DNode:
    def __init__(self, element):
        self.element = element
        self.prev = None
        self.next = None


def insert_between(a, b, value):

    new_node = DNode(value)
    new_node.prev = a
    new_node.next = b
    a.next = new_node
    b.prev = new_node
    return new_node


def delete_node(node):
    if node.prev is not None:
        node.prev.next = node.next
    if node.next is not None:
        node.next.prev = node.prev
    node.prev = None
    node.next = None
````

## D6. Design a Favorites List ADT supporting access(e), remove(e) and top(k). Specify the stored item fields and update behaviour.

**Answer:** A Favorites List tracks how often each element is accessed.

Each stored item contains:
- the element `e`
- an access count

Operations:
- `access(e)`: find the item; if absent, add it with count 0; then increase its count.
- `remove(e)`: remove the item if it exists.
- `top(k)`: return the `k` elements with the highest access counts.

The count is updated on every access. A simple list implementation can sort items by count when `top(k)` is requested.

````python
class FavoritesList:
    def __init__(self):
        self.items = {}  

    def access(self, element):
        if element not in self.items:
            self.items[element] = 0
        self.items[element] += 1

    def remove(self, element):
        if element in self.items:
            del self.items[element]

    def top(self, k):
        if k < 0:
            raise ValueError("k must be non-negative")
        ranked = sorted(self.items.items(),
                        key=lambda item: item[1],
                        reverse=True)
        return [element for element, count in ranked[:k]]


favorites = FavoritesList()
favorites.access("Python")
favorites.access("DSA")
favorites.access("Python")
favorites.access("Colab")
favorites.access("Python")
print(favorites.top(2))
favorites.remove("DSA")
print(favorites.top(2))
````

## D7. Prepare a decision table comparing array-based and linked representations in access, insertion, deletion, memory overhead and worst-case guarantees.

**Answer:**

| Feature | Array-based representation | Linked representation |
|---|---|---|
| Access by index | O(1) | O(n) |
| Search by value | O(n) | O(n) |
| Insert at beginning | O(n), due to shifting | O(1), if head is available |
| Insert at known node/position | Usually O(n) due to shifting | O(1) pointer updates |
| Delete at beginning | O(n), due to shifting | O(1) |
| Memory overhead | Low per element; may have unused capacity | Extra pointer(s) per node |
| Cache performance | Usually better due to contiguous storage | Usually poorer due to separate nodes |
| Worst-case access | O(1) by index | O(n) to reach a node |
| Resizing | Dynamic arrays may occasionally resize in O(n) | No whole-structure resize |

**Conclusion:** Arrays are better for fast indexing and compact storage. Linked lists are useful when frequent insertions/deletions at known positions are required.
